```python
state = 0
LEFT = 0
RIGHT = 1
goal = 2

action = RIGHT

if action == RIGHT:
    if state < 2:
        state = state + 1

if action == LEFT:
    if state > 0:
        state = state - 1

if state == goal:
    reward = 1
else:
    reward = 0

print("State:", state)
print("Reward:", reward)
```

```python
rewards = [0, 0, 1]

total_return = sum(rewards)

print(total_return)
```

```python
rewards = [0, 0, 1]

total_return = 0

for reward in rewards:
    total_return = total_return + reward

print(total_return)
```

```python
rewards = [5, 0, 10]

gamma = 0.9

discounted_return = (
    rewards[0]
    + gamma * rewards[1]
    + (gamma ** 2) * rewards[2]
)

print(discounted_return)
```
