```python
state = 0
LEFT = 0
RIGHT = 1
goal = 2

action = RIGHT

if action == RIGHT:
    if state < 2:
        state = state + 1

if action == LEFT:
    if state > 0:
        state = state - 1

if state == goal:
    reward = 1
else:
    reward = 0

print("State:", state)
print("Reward:", reward)
```
